## 2. Iniciar frontend protegido por senha

A interface gráfica será aberta em um **link temporário de acesso via Gradio**. Esta célula exige definir uma senha forte e utiliza autenticação para proteger o painel.

- O nome de usuário não é um handle de rede social, é apenas um login para este frontend.
- O link e a senha não devem ser divulgados. Encerre a sessão Colab após o uso.
- A aba **Varredura autorizada** faz solicitações públicas apenas quando você confirma a autorização e clica no botão de execução.
- As outras abas são **offline**, adequadas para demonstração com a fictícia Marca Zeta.

Para detalhes dos campos e permissões, leia o [guia de configuração](https://github.com/Ridd1kulusC0d3r/SOSIA/blob/claude/laughing-davinci-bj48ot/docs/CONFIGURATION.md).


```python
#@title Abrir SÓSIA Intelligence Console
ACCESS_USER = "sosia-analyst" #@param {type:"string"}
from getpass import getpass
import gradio as gr
from sosia.webapp import create_app, THEME_CSS

if not ACCESS_USER.strip():
    raise ValueError("Defina um nome de usuário para a interface.")
password = getpass("Senha forte para acessar o frontend (não será exibida): ")
if len(password) < 12:
    raise ValueError("A senha de acesso deve ter pelo menos 12 caracteres.")
app = create_app()
# share=True produz link temporário, protegido por usuário/senha.
app.launch(
    share=True,
    auth=(ACCESS_USER.strip(), password),
    auth_message="SÓSIA é uma interface de investigação autorizada.",
    theme=gr.themes.Soft(primary_hue="teal", neutral_hue="slate"),
    css=THEME_CSS,
    show_error=False,
)
```
